### 머신러닝

- 컴퓨터에게 데이터를 주고, 그 데이터 안에서 규칙(패턴)을 스스로 학습하게 하는 방식
- 학습 종류
    - 지도 학습 => 선형/로지스틱 회귀
    - 비지도 학습 => K-means, PCA
    - 강화 학습

-> ***고전적인 머신러닝***

  ### - 고전적인 머신러닝의 한계
  - 모델이 잘 학습할 수 있도록 사람이 직접 알려줘야함
  - 데이터 양이 일정 수준 넘어서면 복잡한 패턴 학습 못 하고 성능 정체
  - 행/열이 정해진 정형 데이터 처리에 적합, 이미지/텍스트/음성 같은 비정형 데이터 처리 비효율적

### 딥러닝 아이디어
- 인간의 뇌에서 영감 받은 인공신경망 사용해 복잡한 패턴을 학습하는 머신러닝 학습 방법론
- 핵심 아이디어
    - 뇌의 신경망 구조와 동작 방식 컴퓨터로 모방
    - 뇌 속 신경세포(뉴런)들이 신호를 주고 받으며 학습, 판단하듯이 컴퓨터 속 인공 뉴런을 만들어 연결, 네트워크 구축
- 인공 신경망(ANN, Artificial Neural Network)
    - 컴퓨터 안에 인공뉴런(퍼셉트론)을 만들고, 이들을 연결하여 네트워크 구축

### 퍼셉트론

- 여러 정보를 받아서 최종적으로 참(1) 또는 거짓(0)과 같은 결정을 내리는 간단한 결정 모델
- 인공 신경망(ANN)의 가장 기본적 단위로 뉴런의 작동 원리를 모방한 알고리즘
- 구성요소
    - 입력(inputs, x)
    - 가중치(weights, w)
    - 가중합(Weighted Sum)
        - 모든 입력에 각각의 가중치를 곱한 뒤, 모두 더한 값 + 편향
        - (x1+w1) + (x2 + w2) + ... + b 와 같이 "선형 회귀"와 유사
    - 활성화 함수(Activation Function)
    - 입력된 신호를 받아, 그 신호를 처리하여 출력 신호를 결정하는 함수
    - 활성화 시켜 전달할지 말지 결정하는 스위치
  
**퍼셉트론은 데이터 사이에 "직선" 하나를 그어 두 그룹을 나누는 것과 같다**


### 퍼셉트론의 한계 : XOR(배타적 논리합)문제를 해결할 수 없는 문제를 가지고 있다.

- 여러 직선을 쌓자는 아이디어로 신경망구조 등장

### 얕은 신경망(Shallow Neural Network, SNN)

- 초기 인공 신경망의 형태
- 얕은 신경망(Shallow Neural Network)
    - 입력/은닉/출력의 3가지 계층으로 구성
    - 은닉층(Hidden Layer)가 딱 하나만 있는 구조
    - 만능 근사 이론에 의해 이론적으로 모든 문제 해결할 수 있음
- 만능 근사 이론
    - 이론적으로 은닉층에 뉴런이 충분히 많다면
        단 하나의 은닉층만으로 세상의 거의 모든 연속적 함수를 흉내낼 수 있음
    - 하지만 은닉층을 터무니 없이 넓게 만들어야 했고(계산량 매우큼), 단계별/조합적 학습 불가능
- 그래서 뉴런을 넓게 만드는 것이 아니라, 깊게 쌓아올리는 아이디어가 "심층 신경망"

### 심층 신경망(Deep Neural Network, DNN) -> 딥러닝

- 신경망을 옆으로가 아닌 깊게 쌓는 방식
- 여러 개의 은닉층을 가진 인공 신경망
- 얕은 신경망과 비교:
    - 얕은 신경망: 뉴런 12개 => 조합 12개(재사용 불가능)
    - 깊은 신경망 뉴런 6개 * 6개 => 조합 36개
- 더 복잡하고 고차원적인 특징을 학습 가능
- 사람이 특징을 알려주지 않아도, 복잡한 패턴을 스스로 발견

**딥러닝은 심층 신경망 모델을 사용해서 진행**

### 다중 퍼셉트론(Multi-Layer Perceptron, MLP)

- 인공 신경망(Neural Network)의 한 종류로, 퍼셉트론을 여러 층으로 쌓은 모델
- 입력층과 출력층 사이에 1개 이상 은닉층 가짐
- MLP의 학습 과정
  1. 순전파(Forward Propagation)
  2. 손실 함수 계산(Loss Function)
  3. 역전파(Back Propagation) 및 가중치 업데이트(Optimization)
  4. 1~3번 반복 후 종료(Training 완료)

### 순전파(Forward Propagatoin)

-  데이터가 입력층에서 시작해서 여러 은닉층을 거쳐 출력층까지 앞에서 뒤로 흘러가며 계산되는 과정
-  순전파 과정
    1. 초기에 각 뉴런(퍼셉트론)들은 무작위 가중치를 가짐
    2. 데이터(특징 벡터)가 연결된 입력층으로 들어감
    3. 각 뉴런은 입력값과 무작위로 초기화된 가중치를 이용해 계산하고,
        활성화 함수(RELU)를 통과시켜 결과를 다음 층으로 전달
    4. 마지막 출력층까지 반복되어 최정 예측값을 출력
- 초기에는 가중치가 랜덤이기 때문에 결과가 나쁨

### 손실 함수 계산(Loss Function)
- 순전파 과정에서 나온 결과를 실제 정답과 비교해 손실(Loss)/오차(Error)를 구함
- 구하는 방법
    - 회귀(평균 제곱 오차, 각 오차 제곱 합)
    - 분류(교차 엔트로피, 모델이 강하게 확신하고 틀리면 강한 벌점)
    - 비지도(클러스터링 후 떨어진 거리 총합 줄이기)

### 역전파(Back Propagation)

- 손실 값을 가지고, 출력층에서 거꾸로 계산하면서 가중치를 업데이트하는 과정
- 역전파 과정
  1. 발생한 오차를 이전 층으로 거꾸로 추적하면서 각 가중치가 오차에 준 영향 구함
  2. 계산된 영향력을 바탕으로 오차를 줄이는 방향으로 각 가중치 업데이트
  3. 첫 은닉층까지 도달 후 순전파 반복

### Training 완료

순전파 -> 오차 계산 -> 역전파 과정을 언제까지 진행할까?

1. 정해둔 학습 횟수(Epoch)에 도달
   - 언제 끝날지 예측 가능하고 구현 간단
   - 너무 적게 반복하면 과소적합, 너무 많이 반복하면 과대적합
2. 조기종료 조건 걸기

MLP 성능 향상시키는 기술들
- 우리가 배운 기본 MLP는 실제 복잡한 환경에서 제대로 동작하지 않음
- 대표적인 향상 기술
  1. 더 좋은 활성화 함수(ReLU)
  2. 더 효율적인 학습 위한 옵티마이저

### 활성화 함수
- 입력된 신호를 받아, 그 신호를 처리해 출력 신호를 결정하는 함수
- "활성화"시켜 전달할지 말지 결정하는 스위치
- 활성화 함수 필요한 이유
    - 활성화 함수 없다면, 신경망은 여러 층을 아무리 깊게 쌓아도 결국 거대한 선형 함수
    - 활성화 함수는 이런 선형 함수가 훨씬 복잡한 패턴을 학습할 수 있게 함

초기에 사용되던 시그모이드 함수는 값을 0~1 사이로 지정해 좋아보였으나, 역전파 과정에서 "기울기 소실 문제" 발생

**기울기 소실 문제(Vanishing Gradient Problem)**
- 시그모이드는 역전파 과정에서 오차 신호를 전달할때 세기를 약하게 만듦
- 층을 거스를때마다 작은 값들이 계속 곱해짐
- 입력층 근처로 갈수록 0에 수렴(Vanishing)
- 앞쪽 층 뉴런들은 오차 피드백을 받지 못해 학습 정지

이를 해결하기 위한 것이 ReLU 활성화 함수

### ReLU 활성화 함수
- Rectified Linear Unit
- 입력값이 0보다 작으면 0, 크면 그대로 출력

1. 기울기 소실 문제 해결
    - 입력값 0보다 클 때 기울기는 무조건 1
    - 역전파 과정에서 신호 세기 줄어들지 않고 앞쪽 층까지 온전히 전달
    - but 음수만 지속적으로 들어오는 경우 DeadLeLU 문제 발생(죽은 뉴런), 이를 해결하기 위해 Leaky ReLU 활용
2. 매우 빠른 계산 속도
    - 복잡한 지수 함수 계산 필요 없음

## 옵티마이저

### Adam

1. 관성 도입
   - 움직였던 거리와 방향을 기억
   
2. 적응적 학습률 도입
    - 가중치마다 "고유한 학습률"을 부여해 최적의 학습 속도를 유지해 효율적으로 낮은 손실 지점 도달
    - 많이 움직였던 가중치에 대해 학습률 적게, 적게 이동
    - 적게 움직였던 가중치에 대해 학습률 크게, 많이 이동

## 규제

- 모델의 과적합을 막기 위해 제약을 거는 것
- 대표적인 규제 방법
  1. L1 라쏘
    - 일부 가중치를 0으로 만들어 제거 -> 복잡도 감소 / 불필요 특성 많을 시 사용
  2. L2 릿지
    - 가중치를 0에 가깝게 만들지만 완전히 0은 아님
    - 다중공산성 있을 때, 상관관계 높은 여러 특성들에 걸쳐 영향력 고르게 분산
  3. 드롭아웃
    - 학습 시 각 뉴런을 랜덤하게 일시적으로 학습시키지 않음
