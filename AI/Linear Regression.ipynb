## 1-1.선형회귀(Linear Regression)

```선형회귀란?```
- 입력 변수와 출력 변수 사이의 관계를 직선 형태로 근사해 예측하는 통계적 방법
- 지도학습의 가장 ***기초***가 되는 접근법
- 단순해 보이지만 개념적으로도, 실무적으로도 매우 유용

## 2-1. 단순선형회귀: 단일 변수를 이용한 선형회귀

```단순선형회귀(simple linear regression)란?```
- 한 개의 설명변수(X)와 하나의 반응변수(Y) 사이의 선형(직선) 관계를 찾는 방법
- 목표: 데이터를 가장 잘 설명하는 직선을 찾아 예측에(y hat) 활용
- 단일 설명변수를 이용한 단순선형회귀
    - 모형 가정: Y = a + bX + E(입실론, 상실함수)
        - a: 절편(X = 0일때 Y값)
        - b: 기울기(X가 1 단위 증가할 때 Y의 평균 증가량)
        - E: 관측 오차
    - hat 표기는 추정값을 의미

## 2-2. 최소제곱법(least squares)
```최소제곱법(least squares)이란?```
- ***실제 관측값과 예측값의 차이***(잔차, residual)를 제곱해 합한 값(RSS, **잔차제곱합**)을 최소화하는 방법
- 목표: 데이터를 가장 잘 설명하는 직선을 찾기 위해 계수 a, b를 추정

잔차: y - (y hat)

예측값(y hat) = a + bx

RSS(잔차제곱합): 잔차 제곱의 합

RSS = 시그마[(y-a-bx) ** 2]

기울기 = 시그마[(xi-x.mean())(yi-y.mean())] / 시그마[(xi-x.mean()) ** 2]
